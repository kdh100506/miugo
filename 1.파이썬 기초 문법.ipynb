# 1. 변수와 데이터 타입

## 1-1. 변수란?

변수는 **값을 저장하는 이름표**입니다.  
프로그램에서 숫자, 문자, 계산 결과 등을 저장해 두고 다시 사용할 수 있습니다.


````python
name = "민수"
age = 17

print(name)
print(age)
````

## 1-2. 변수 이름 규칙

변수 이름을 만들 때는 몇 가지 규칙이 있습니다.

| 규칙 | 예시 |
|---|---|
| 대문자, 소문자 구분함 | 'name'과 'NAME'은 다른 변수 |
| 영어, 숫자, 밑줄 `_` 사용 가능 | `student_name`, `score1` |
| 숫자로 시작할 수 없음 | `1score`는 오류 |
| 공백을 넣을 수 없음 | `student name`은 오류 |
| 파이썬 예약어는 사용할 수 없음 | `if`, `for`, `while` 등 |

좋은 변수 이름 예시입니다.


````python
student_name = "지훈"
math_score = 90
is_pass = True
````

좋지 않은 변수 이름 예시입니다.


````python
a = "지훈"
b = 90
c = True
````

### 헷갈리기 쉬운 부분 ① 대입 `=`과 비교 `==`

| 기호 | 역할 | 예시 |
|---|---|---|
| `=` | 오른쪽 값을 왼쪽 변수에 저장 | `score = 80` |
| `==` | 두 값이 같은지 비교 | `score == 80` |

조건문에서 `score = 80`처럼 작성하면 비교가 아니라 대입이므로 올바른 조건식이 아닙니다.

또한 변수는 **사용하기 전에 반드시 값을 저장**해야 합니다. 값을 저장하지 않은 변수를 사용하면 `NameError`가 발생합니다.


````python
score = 80              # 값을 저장
is_eighty = score == 80 # 같은지 비교

print(score)
print(is_eighty)
````

### 심화 학습 ① 여러 변수에 값 저장하기

파이썬은 여러 변수에 한 번에 값을 저장하거나 두 변수의 값을 간단히 바꿀 수 있습니다.


````python
name, age = "지훈", 17
print(name, age)

left = "왼쪽"
right = "오른쪽"
left, right = right, left

print(left, right)
````

### 변수 이름 추가 주의사항

- `studentName`보다 `student_name`처럼 단어 사이에 밑줄을 사용하는 방식이 파이썬에서 일반적입니다.
- 대문자로만 작성한 이름은 보통 바뀌지 않는 상수처럼 사용합니다. 예: `MAX_SCORE = 100`
- 파이썬 공식 스타일 지침인 **Python Enhancement Proposal 8(PEP 8)**에서는 의미가 분명한 소문자 변수 이름을 권장합니다.
- `list`, `str`, `sum`처럼 파이썬이 이미 제공하는 함수 이름을 변수 이름으로 사용하지 않는 것이 좋습니다.

```python
# 권장하지 않는 예
list = [1, 2, 3]  # 이후 list() 함수를 사용하기 어려워짐
```


변수 이름은 가능하면 **무엇을 저장하는지 알 수 있게** 작성하는 것이 좋습니다.

## 1-3. 데이터 타입이란?

데이터 타입은 **값의 종류**를 의미합니다.

파이썬에서 자주 사용하는 기본 데이터 타입은 다음과 같습니다.

| 데이터 타입 | 설명 | 예시 |

| `bool` | 참 또는 거짓 | `True`, `False` |

## 1-4. 정수형 `int`

정수형은 소수점이 없는 숫자입니다.


````python
age = 17
count = 10

print(age)
print(count)
````

### 심화 학습 ② 정수 표현

큰 정수는 밑줄 `_`을 넣어 읽기 쉽게 작성할 수 있습니다. 밑줄은 값에 영향을 주지 않습니다.


````python
population = 51_000_000
budget = 1_250_000_000

print(population)
print(budget)
print(1_000 == 1000)
````

## 1-5. 실수형 `float`

실수형은 소수점이 있는 숫자입니다.


````python
height = 175.5
temperature = 36.5

print(height)
print(temperature)
````

### 헷갈리기 쉬운 부분 ② 실수의 오차

컴퓨터는 일부 소수를 정확하게 저장하지 못합니다. 따라서 실수 계산 결과를 `==`로 직접 비교하면 예상과 다른 결과가 나올 수 있습니다.


````python
result = 0.1 + 0.2

print(result)
print(result == 0.3)
print(round(result, 1) == 0.3)
````

정확한 비교가 필요하면 `round()`로 자릿수를 맞추거나 `math.isclose()`를 사용할 수 있습니다.


````python
import math

result = 0.1 + 0.2
print(math.isclose(result, 0.3))
````

## 1-6. 문자열 `str`

문자열은 글자 데이터를 의미합니다.  
작은따옴표 또는 큰따옴표로 감싸서 사용합니다.


````python
school = "대구소프트웨어마이스터고"
subject = 'Python'

print(school)
print(subject)
````

문자열끼리는 더할 수 있습니다.


````python
first_name = "김"
last_name = "민수"

full_name = first_name + last_name

print(full_name)
````

### 헷갈리기 쉬운 부분 ③ 문자열과 숫자는 바로 더할 수 없음

`"나이: " + 17`처럼 문자열과 정수를 `+`로 연결하면 `TypeError`가 발생합니다. 숫자를 문자열로 바꾸거나 f-string을 사용해야 합니다.


````python
name = "민수"
age = 17

print("이름: " + name + ", 나이: " + str(age))
print(f"이름: {name}, 나이: {age}")
````

### 심화 학습 ③ f-string

f-string은 문자열 앞에 `f`를 붙이고 중괄호 `{}` 안에 변수나 수식을 넣는 방법입니다.


````python
price = 12_000
quantity = 3

print(f"가격: {price:,}원")
print(f"수량: {quantity}개")
print(f"합계: {price * quantity:,}원")
````

### 심화 학습 ④ 문자열 인덱싱과 슬라이싱

문자열의 위치 번호는 **0부터 시작**합니다. 슬라이싱의 끝 위치는 포함되지 않습니다.


````python
word = "Python"

print(word[0])      # 첫 글자
print(word[-1])     # 마지막 글자
print(word[0:3])    # 0, 1, 2 위치
print(word[2:])     # 2 위치부터 끝까지
print(word[::-1])   # 거꾸로
````

문자열은 생성된 뒤 특정 글자만 직접 바꿀 수 없는 **불변(immutable) 자료형**입니다.

```python
word = "Python"
word[0] = "J"  # TypeError 발생
```

새 문자열을 만들어 변수에 다시 저장해야 합니다.


````python
word = "Python"
word = "J" + word[1:]
print(word)
````

## 1-7. 불 자료형 `bool`

불 자료형은 참과 거짓을 나타냅니다.


````python
is_student = True
is_adult = False

print(is_student)
print(is_adult)
````

조건문에서 자주 사용됩니다.


````python
score = 85
is_pass = score >= 60

print(is_pass)
````

### 심화 학습 ⑤ 참처럼 판단되는 값과 거짓처럼 판단되는 값

조건문에는 `True`, `False`뿐 아니라 다른 값도 사용할 수 있습니다.

| 거짓으로 판단되는 대표 값 | 참으로 판단되는 대표 값 |
|---|---|
| `0`, `0.0` | 0이 아닌 숫자 |
| 빈 문자열 `""` | 내용이 있는 문자열 |
| 빈 리스트 `[]` | 값이 있는 리스트 |
| `None` | 대부분의 일반 객체 |


````python
values = [0, 1, "", "Python", [], [1], None]

for value in values:
    print(f"{value!r:10} -> {bool(value)}")
````

조건식에 불필요하게 `== True`를 붙이지 않는 것이 일반적입니다.

```python
# 권장

    print("학생입니다.")

if is_student == True:
```

## 1-8. 데이터 타입 확인하기

`type()` 함수를 사용하면 값의 데이터 타입을 확인할 수 있습니다.


````python
name = "민수"
age = 17
height = 175.5
is_student = True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
````

### 심화 학습 ⑥ `isinstance()`로 자료형 확인하기

`type()`은 정확한 자료형을 확인할 때 사용하고, `isinstance()`는 특정 자료형에 해당하는지 참·거짓으로 판단할 때 편리합니다.


````python
age = 17
height = 175.5

print(isinstance(age, int))
print(isinstance(height, float))
print(isinstance(age, (int, float)))
````

### 헷갈리기 쉬운 부분 ④ 파이썬 변수의 자료형은 바뀔 수 있음

파이썬은 변수에 저장되는 값에 따라 자료형이 결정됩니다. 이를 **동적 타이핑(dynamic typing)**이라고 합니다.


````python
data = 10
print(data, type(data))

data = "10"
print(data, type(data))
````

자료형을 바꾸어 저장할 수는 있지만, 같은 변수가 서로 전혀 다른 의미의 값을 번갈아 저장하면 코드를 이해하기 어려워집니다.

## 1-9. 데이터 타입 변환(형변환:Casting)

입력받은 값이나 저장된 값을 다른 데이터 타입으로 바꿀 수 있습니다.

| 함수 | 설명 |
|---|---|
| `int()` | 정수로 변환 |
| `float()` | 실수로 변환 |
| `str()` | 문자열로 변환 |

예시입니다.


````python
number = "10"

print(number + "5")
print(int(number) + 5)
````

`input()`으로 입력받은 값은 기본적으로 문자열입니다.


````python
age = input("나이를 입력하세요: ")

print(type(age))
````

숫자로 계산하려면 `int()`로 변환해야 합니다.


````python
age = int(input("나이를 입력하세요: "))

next_year = age + 1

print("내년 나이:", next_year)
````

### 헷갈리기 쉬운 부분 ⑤ 형 변환 오류

- `int("10")`은 가능하지만 `int("3.14")`는 바로 변환할 수 없습니다.
- 숫자가 아닌 문자열을 `int()`로 바꾸면 `ValueError`가 발생합니다.
- 사용자 입력 앞뒤의 불필요한 공백은 `strip()`으로 제거할 수 있습니다.


````python
text = "  42  "
number = int(text.strip())
print(number)

float_text = "3.14"
number = int(float(float_text))
print(number)  # 소수점 이하는 버려짐
````

### 심화 학습 ⑦ 변환 오류 처리하기

`try-except`를 사용하면 잘못된 값이 들어와도 프로그램이 갑자기 종료되지 않도록 처리할 수 있습니다.


````python
text = "python"

try:
    number = int(text)
    print(number)
except ValueError:
    print("정수로 바꿀 수 없는 값입니다.")
````

> `int(3.9)`의 결과는 반올림한 `4`가 아니라 소수점 이하를 버린 `3`입니다. 반올림은 `round()`를 사용합니다.


````python
print(int(3.9))
print(round(3.9))
````

## 2-1. 수식이란?

## 2-2. 산술 연산자

산술 연산자는 숫자를 계산할 때 사용합니다.

| 연산자 | 의미 | 예시 | 결과 |
|---|---|---|---|
| `+` | 더하기 | `10 + 3` | `13` |
| `-` | 빼기 | `10 - 3` | `7` |
| `*` | 곱하기 | `10 * 3` | `30` |
| `/` | 나누기 | `10 / 3` | `3.333...` |
| `//` | 몫 | `10 // 3` | `3` |
| `%` | 나머지 | `10 % 3` | `1` |
| `**` | 거듭제곱 | `2 ** 3` | `8` |

예시 코드입니다.


````python
a = 10
b = 2

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** b)
````

### 헷갈리기 쉬운 부분 ⑥ `/`와 `//`

- `/`는 나누기이며 결과가 항상 실수형 `float`입니다.
- `//`는 소수점 이하를 단순히 잘라내는 연산이 아니라 **아래쪽 정수로 내림**하는 연산입니다.


````python
print(6 / 3, type(6 / 3))
print(7 // 3)
print(-7 // 3)  # -2가 아니라 -3
print(-7 % 3)   # 몫과 나머지의 관계 확인
````

항상 다음 관계가 성립합니다.

```text
나누어지는 수 = 나누는 수 × 몫 + 나머지
```


````python
a = -7
b = 3

quotient, remainder = divmod(a, b)
print(quotient, remainder)
print(a == b * quotient + remainder)
````

나머지 연산자는 짝수, 홀수를 판별할 때 자주 사용합니다.


````python
number = 7

if number % 2 == 0:
    print("짝수입니다.")
else:
    print("홀수입니다.")
````

나머지 연산자는 짝수·홀수 판별뿐 아니라 값이 일정 범위 안에서 반복되도록 만들 때도 사용합니다.


````python
# 0, 1, 2가 반복됨
for number in range(8):
    print(number, "->", number % 3)
````

## 2-4. 비교 연산자

비교 연산자는 두 값을 비교할 때 사용합니다.  
결과는 `True` 또는 `False`입니다.

| 연산자 | 의미 |
|---|---|
| `==` | 같다 |
| `!=` | 같지 않다 |
| `>` | 크다 |
| `<` | 작다 |
| `>=` | 크거나 같다 |
| `<=` | 작거나 같다 |

예시 코드입니다.


````python
score = 85

print(score >= 60)
print(score == 100)
print(score != 0)
````

### 헷갈리기 쉬운 부분 ⑦ `==`와 `is`

- `==`는 **값이 같은지** 비교합니다.
- `is`는 **같은 객체인지** 비교합니다.
- 일반적인 숫자·문자열 비교에는 `==`를 사용합니다.
- 값이 없음을 나타내는 `None`은 보통 `is None` 또는 `is not None`으로 확인합니다.


````python
value = None

print(value is None)
print(value is not None)
````

### 심화 학습 ⑨ 연속 비교

파이썬은 수학식처럼 비교 연산자를 연결할 수 있습니다.


````python
score = 85

print(0 <= score <= 100)
print(score >= 0 and score <= 100)
````

## 2-5. 논리 연산자

논리 연산자는 여러 조건을 함께 판단할 때 사용합니다.

| 연산자 | 의미 |
|---|---|
| `and` | 두 조건이 모두 참이면 참 |
| `or` | 두 조건 중 하나라도 참이면 참 |
| `not` | 참과 거짓을 반대로 바꿈 |

예시 코드입니다.


````python
score = 85
attendance = 90

if score >= 60 and attendance >= 80:
    print("통과입니다.")
else:
    print("재시험입니다.")
````

`or` 예시입니다.


````python
has_coupon = True
is_member = False

if has_coupon or is_member:
    print("할인을 받을 수 있습니다.")
else:
    print("할인을 받을 수 없습니다.")
````

### 심화 학습 ⑩ 단락 평가(short-circuit evaluation)

`and`와 `or`는 결과가 이미 결정되면 뒤의 조건을 검사하지 않습니다.

- `A and B`: A가 거짓이면 B를 검사하지 않음
- `A or B`: A가 참이면 B를 검사하지 않음

이 성질을 이용하면 오류가 발생할 수 있는 조건을 안전하게 검사할 수 있습니다.


````python
text = ""

# text가 빈 문자열이면 text[0]을 검사하지 않으므로 오류가 나지 않음
if text and text[0] == "A":
    print("A로 시작합니다.")
else:
    print("A로 시작하지 않거나 빈 문자열입니다.")
````

### 헷갈리기 쉬운 부분 ⑧ `not`의 범위

복잡한 조건은 괄호를 사용하여 의도를 분명히 하는 것이 좋습니다.


````python
age = 17
has_permission = False

print(not age >= 18)
print(not (age >= 18))
print(not (age >= 18 and has_permission))
````

## 3-3. `if`, `elif`, `else`

조건이 여러 개일 때는 `elif`를 사용합니다.


````python
score = 85

if score >= 90:
    print("A등급")
elif score >= 80:
    print("B등급")
elif score >= 70:
    print("C등급")
else:
    print("D등급")
````

### 헷갈리기 쉬운 부분 ⑪ `elif`는 위에서부터 검사함

처음으로 참이 된 블록 하나만 실행됩니다. 따라서 범위가 넓은 조건을 먼저 작성하면 아래의 세부 조건이 실행되지 않을 수 있습니다.


````python
score = 95

# 잘못된 순서의 예
if score >= 60:
    print("합격")
elif score >= 90:
    print("A등급")
````

위 코드에서 95점은 첫 번째 조건 `score >= 60`이 이미 참이므로 `A등급`을 검사하지 않습니다. 등급처럼 범위를 나눌 때는 일반적으로 **큰 값부터 작은 값 순서**로 검사합니다.

경계값인 59, 60, 69, 70, 79, 80, 89, 90 등을 직접 테스트하면 조건식 오류를 쉽게 발견할 수 있습니다.


### 심화 학습 ⑪ 조건 표현식(삼항연산)

간단한 `if-else`는 한 줄의 조건 표현식으로 작성할 수 있습니다.


````python
score = 75
result = "합격" if score >= 60 else "불합격"
print(result)
````

### match-case문

`match-case`문은 하나의 값을 여러 경우와 비교하여 일치하는 코드를 실행하는 조건문입니다.
메뉴 번호나 명령어처럼 **정해진 값에 따라 기능을 나눌 때** 유용합니다.

* `match`: 비교할 값을 작성합니다.
* `case`: 일치하는 값을 작성합니다.
* `case _`: 일치하는 값이 없을 때 실행되며, `else`와 비슷합니다.
* 파이썬 3.10 이상에서 사용할 수 있습니다.
* java 처럼 break구문이 필요없습니다.

````python
menu = int(input("메뉴를 선택하세요(1~3): "))

match menu:
    case 1:
        print("학생을 등록합니다.")
    case 2:
        print("학생을 조회합니다.")
    case 3:
        print("학생을 삭제합니다.")
    case _:
        print("잘못된 메뉴입니다.")
````

조건문 안에 또 다른 조건문을 넣을 수 있습니다.


````python
score = 85
attendance = 95

if score >= 60:
    if attendance >= 80:
        print("최종 통과입니다.")
    else:
        print("출석 부족입니다.")
else:
    print("점수 부족입니다.")
````

### 심화 학습 ⑫ 중첩을 줄여 읽기 쉽게 만들기

두 조건을 모두 만족해야 한다면 중첩 조건문 대신 `and`로 합칠 수 있습니다.


````python
score = 85
attendance = 95

if score >= 60 and attendance >= 80:
    print("최종 통과입니다.")
elif score < 60:
    print("점수 부족입니다.")
else:
    print("출석 부족입니다.")
````

다만 조건이 너무 많이 중첩되면 코드가 복잡해질 수 있습니다.

````python
total = 0

for i in range(1, 11):
    total += i

print("합계:", total)
````

### 심화 학습 ⑭ 합계를 구하는 여러 가지 방법

반복문은 계산 과정을 이해하는 데 좋고, 실제 코드에서는 내장 함수 `sum()`이 더 간결합니다. 1부터 n까지의 합은 공식으로도 구할 수 있습니다.


````python
n = 10

loop_total = 0
for i in range(1, n + 1):
    loop_total += i

sum_total = sum(range(1, n + 1))
formula_total = n * (n + 1) // 2

print(loop_total, sum_total, formula_total)
assert loop_total == sum_total == formula_total
````

`assert`는 조건이 참인지 확인하는 문장입니다. 조건이 거짓이면 `AssertionError`가 발생하므로 계산 결과나 프로그램 상태를 점검할 때 사용할 수 있습니다.

## 3-9. `while` 반복문

`while`문은 조건이 참인 동안 계속 반복합니다.


````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

### 헷갈리기 쉬운 부분 ⑬ `while`문의 상태 변경

조건에 사용한 값이 반복문 안에서 바뀌지 않으면 무한 반복이 발생할 수 있습니다.

```python
# 무한 반복 예시
count = 1
while count <= 5:
    print(count)
    # count += 1이 없음
```

반복 조건, 초기값, 상태 변경 위치를 함께 확인해야 합니다.


````python
count = 1

while count <= 5:
    print(count)
    count += 1

print("반복 종료 후 count:", count)
````

`while`문은 조건을 잘못 작성하면 무한 반복이 될 수 있으므로 주의해야 합니다.

## 3-10. `break`

`break`는 반복문을 즉시 종료할 때 사용합니다.


````python
while True:
    password = input("비밀번호 입력: ")

    if password == "1234":
        print("로그인 성공")
        break
    else:
        print("비밀번호가 틀렸습니다.")
````

`break`는 가장 가까운 반복문 하나만 종료합니다. 중첩 반복문 전체를 한 번에 종료하는 문장은 아니므로 주의해야 합니다.

## 3-11. `continue`

`continue`는 반복문의 나머지 부분을 건너뛰고 다음 반복으로 넘어갈 때 사용합니다.


````python
for i in range(1, 11):
    if i % 2 == 0:
        continue

    print(i)
````

위 코드는 짝수일 때 `continue`를 만나 출력하지 않고 넘어갑니다.

### 헷갈리기 쉬운 부분 ⑭ `while`문에서 `continue`

`continue`보다 아래에 상태 변경 코드를 두면 그 코드가 실행되지 않아 무한 반복이 될 수 있습니다. 상태 변경을 먼저 수행해야 합니다.


````python
count = 0

while count < 5:
    
    if count == 3:
        continue

    count += 1
    
    print(count)
````

## 문제 8. 로그인 기회 제한하기

비밀번호를 최대 3번까지 입력받습니다.

- 비밀번호가 `"python123"`이면 `"로그인 성공"`을 출력하고 종료합니다.
- 3번 모두 틀리면 `"로그인 실패"`를 출력합니다.


````python
chance = 0

while True:
  if chance == 3:
    print('로그인 실패')
    break
  password = input('비밀번호')
  if password == 'python123':
    print('로그인 성공')
    break
  else:
    print('비밀번호가 틀렸습니다. 다시 입력해주세요.')
    chance += 1
    continue
````

## 문제 9. 3과 5의 배수 출력하기

1부터 30까지 반복하면서 다음 규칙대로 출력하세요.

- 3과 5의 공배수: `FizzBuzz`
- 3의 배수: `Fizz`
- 5의 배수: `Buzz`
- 나머지 숫자: 숫자 그대로 출력

> 공배수 조건을 가장 먼저 검사해야 합니다.


````python
FizzBuzz = []
Fizz = []
Buzz = []

for i in range(1,31):
  if i % 3 == 0 and i % 5 == 0:
    FizzBuzz.append(i)
    Fizz.append(i)
    Buzz.append(i)
  elif i % 3 == 0:
    Fizz.append(i)
  elif i % 5 == 0:
    Buzz.append(i)

print(FizzBuzz)
print(Fizz)
print(Buzz)
````

## 문제 10. 점수 목록 분석하기

다음 점수 목록을 사용하여 아래 내용을 출력하세요.

```python
scores = [85, 42, 77, 90, 58, 100, 63]
```

1. 평균 점수
2. 60점 이상인 점수만 담은 새 리스트
3. 최고 점수와 최저 점수
4. 합격 인원 수


````python
scores = [85, 42, 77, 90, 58, 100, 63]

avg = sum(scores)/len(scores)
up_60 = []
for i in scores:
  if i >= 60:
    up_60.append(i)
max_score = max(scores)
min_score = min(scores)

print(f'평균 : {avg}점')
print(f'60점 이상 : {up_60}')
print(f'최고 점수, 최저 점수 : {max_score}점, {min_score}점')
print(f'합격 인원수 : {len(up_60)}명')
````
