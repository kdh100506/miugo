### 1. 파일 읽기 관련 메소드 (Reading)

| 메소드명 | 역할 | 리턴 값 타입 | 특징 및 주의점 |
| :--- | :--- | :--- | :--- |
| **`file.read(size=-1)`** | 파일의 내용 전체(또는 지정한 크기만큼)를 한 번에 읽어옵니다. | `str` (또는 `bytes`) | 파일이 너무 크면 메모리가 부족해질 수 있습니다. |
| **`file.readline()`** | 파일에서 **딱 한 줄**만 읽어옵니다. | `str` | 줄바꿈 문자(`\n`)를 포함하여 읽으며, 파일 끝에 도달하면 빈 문자열(`""`)을 리턴합니다. |
| **`file.readlines()`** | 파일의 모든 줄을 읽어서 **리스트**로 만듭니다. | `list` | 각 줄이 리스트의 요소(`['1줄\n', '2줄\n']`)로 저장됩니다. |

<br>

### 2. 파일 쓰기 관련 메소드 (Writing)

| 메소드명 | 역할 | 리턴 값 타입 | 특징 및 주의점 |
| :--- | :--- | :--- | :--- |
| **`file.write(string)`** | 파일에 문자열을 기록합니다. | `int` (글자 수) | 자동 줄바꿈이 되지 않으므로, 줄을 바꾸려면 문자열 끝에 직접 **`\n`**을 더해야 합니다. |
| **`file.writelines(lines)`** | 문자열이 담긴 리스트를 한 번에 파일에 씁니다. | `None` | 리스트 내의 문자열들에 줄바꿈 문자(`\n`)가 포함되어 있어야 줄별로 저장됩니다. |

<br>

### 3. 파일 상태 및 제어 관련 메소드 (Status & Control)

| 메소드명 | 역할 | 리턴 값 타입 | 특징 및 주의점 |
| :--- | :--- | :--- | :--- |
| **`file.close()`** | 열려 있는 파일을 닫고 시스템 자원을 반납합니다. | `None` | `with` 구문을 사용하면 내부적으로 자동 호출되므로 직접 쓸 일이 적습니다. |
| **`file.flush()`** | 버퍼에 남아있는 출력 대기 데이터를 파일에 즉시 강제로 씁니다. | `None` | 파일을 닫지 않고도 중간에 데이터를 물리적 디스크에 기록하고 싶을 때 씁니다. |
| **`file.tell()`** | 파일 내부에서 현재 읽기/쓰기가 진행 중인 **포인터(커서)의 위치**를 알려줍니다. | `int` (바이트 단위) | 파일의 시작 지점(0)부터 현재 위치까지의 거리를 반환합니다. |
| **`file.seek(offset, whence=0)`** | 파일 포인터(커서)의 위치를 원하는 곳으로 **이동**시킵니다. | `int` (새 포인터 위치) | 파일을 처음부터 다시 읽고 싶을 때 `file.seek(0)` 형태로 자주 활용합니다. |

# 2. 파일 쓰기 기본

파일에 내용을 저장할 때는 `open()` 함수를 사용합니다.

```python
file = open("파일명", "w", encoding="utf-8")
file.write("저장할 내용")
file.close()
```

| 코드 | 의미 |
|---|---|
| `"w"` | 쓰기 모드 |
| `encoding="utf-8"` | 한글이 깨지지 않도록 설정 |
| `write()` | 파일에 내용 쓰기 |
| `close()` | 파일 닫기 |


````python
file = open("hello.txt", "w", encoding="utf-8")
file.write("안녕하세요. 파이썬 파일 입 출력입니다.")
file.close()

print("파일 저장 완료")
````

# 4. `with open()` 사용하기

파일을 열면 마지막에 닫아야 합니다.  
`with open()`을 사용하면 파일을 자동으로 닫아줍니다.

파일 입출력에서는 가능하면 `with open()`을 사용하는 것이 좋습니다.


````python
with open("hello_with.txt", "w", encoding="utf-8") as file:
    file.write("with open을 사용하면 파일을 자동으로 닫습니다.")

print("파일 저장 완료")
````

# 5. 파일 모드

| 모드 | 의미 | 설명 |
|---|---|---|
| `r` | read | 읽기 모드 |
| `w` | write | 쓰기 모드, 기존 내용 덮어쓰기 |
| `a` | append | 추가 모드, 기존 내용 뒤에 추가 |


````python
with open("log.txt", "w", encoding="utf-8") as file:
    file.write("첫 번째 기록\n")

with open("log.txt", "a", encoding="utf-8") as file:
    file.write("두 번째 기록\n")

with open("log.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

# 14. 딕셔너리 리스트를 텍스트 파일에 저장하기

프로젝트에서는 리스트 안에 딕셔너리를 넣어 데이터를 관리하는 경우가 많습니다.


````python
students = [
    {"name": "민수", "score": 90},
    {"name": "지훈", "score": 80},
    {"name": "서준", "score": 95}
]

with open("student_scores.txt", "w", encoding="utf-8") as file:
    for student in students:
        file.write(student["name"] + "," + str(student["score"]) + "\n")

print("학생 점수 저장 완료")
````

````python
students = []

with open("student_scores.txt", "r", encoding="utf-8") as file:
    for line in file:
        name, score = line.strip().split(",")

        student = {
            "name": name,
            "score": int(score)
        }

        students.append(student)

print(students)
````

# 15. CSV 파일이란?

Comma-Separated Values(CSV, 쉼표로 구분된 표 형식 데이터)는 데이터를 쉼표로 구분해서 저장하는 파일 형식입니다.

```text

서준,95
```

CSV 파일은 엑셀에서도 열 수 있고, 파이썬에서도 쉽게 읽고 쓸 수 있습니다.

# 17. CSV 파일 쓰기: `csv.writer`

`csv.writer`를 사용하면 리스트 형태의 데이터를 CSV 파일에 저장할 수 있습니다.


````python
import csv

with open("students.csv", "w", encoding="utf-8", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["name", "score"])
    writer.writerow(["민수", 90])
    writer.writerow(["지훈", 80])
    writer.writerow(["서준", 95])

print("CSV 파일 저장 완료")
````

# 19. CSV 헤더 건너뛰기

CSV 파일의 첫 줄은 보통 컬럼 이름입니다.  
이를 헤더라고 합니다.

`next()`를 사용하면 첫 줄을 건너뛸 수 있습니다.


````python
import csv

with open("students.csv", "r", encoding="utf-8") as file:
    reader = csv.reader(file)

    header = next(reader)
    print("헤더:", header)

    for row in reader:
        name = row[0]
        score = int(row[1])
        print(name, score)
````

# 21. 게임 랭킹 데이터를 CSV 파일에 저장하기

# 22. `csv.DictWriter`

`csv.DictWriter`를 사용하면 딕셔너리 데이터를 CSV 파일에 저장하기 쉽습니다.


````python
import csv

students = [
    {"name": "민수", "score": 90},
    {"name": "지훈", "score": 80},
    {"name": "서준", "score": 95}
]

with open("students_dict.csv", "w", encoding="utf-8", newline="") as file:
    fieldnames = ["name", "score"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(students)

print("DictWriter로 저장 완료")
````

# 24. 함수로 저장 기능 만들기

프로젝트에서는 저장 기능을 함수로 분리하는 것이 좋습니다.


````python
import csv

def save_students(students, filename):
    with open(filename, "w", encoding="utf-8", newline="") as file:
        writer = csv.writer(file)

        writer.writerow(["name", "score"])

        for student in students:
            writer.writerow([student["name"], student["score"]])

students = [
    {"name": "민수", "score": 90},
    {"name": "지훈", "score": 80}
]

save_students(students, "students_save.csv")

print("저장 완료")
````

# 25. 함수로 불러오기 기능 만들기


````python
import csv

def load_students(filename):
    students = []

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        next(reader)

        for row in reader:
            student = {
                "name": row[0],
                "score": int(row[1])
            }

            students.append(student)

    return students

students = load_students("students_save.csv")

print(students)
````

# 27. 프로젝트용 예제: 게임 랭킹 저장과 불러오기


````python
import csv

class GameRecord:
    def __init__(self, nickname, game_name, score):
        self.nickname = nickname
        self.game_name = game_name
        self.score = score

    def show_info(self):
        print(self.nickname, self.game_name, self.score)

def save_records(records, filename):
    with open(filename, "w", encoding="utf-8", newline="") as file:
        writer = csv.writer(file)

        writer.writerow(["nickname", "game_name", "score"])

        for record in records:
            writer.writerow([record.nickname, record.game_name, record.score])

def load_records(filename):
    records = []

    with open(filename, "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        next(reader)

        for row in reader:
            record = GameRecord(row[0], row[1], int(row[2]))
            records.append(record)

    return records

records = [
    GameRecord("dragon", "테트리스", 1200),
    GameRecord("tiger", "테트리스", 1500),
    GameRecord("lion", "테트리스", 900)
]

save_records(records, "game_ranking.csv")

loaded_records = load_records("game_ranking.csv")

for record in loaded_records:
    record.show_info()
````

# 29. 예외 처리가 필요한 이유

프로그램을 실행하다 보면 예상하지 못한 오류가 발생할 수 있습니다.

예를 들어 다음과 같은 상황입니다.

| 상황 | 발생할 수 있는 오류 |
|---|---|
| 없는 파일을 읽으려고 할 때 | 파일을 찾을 수 없음 |
| 숫자를 입력해야 하는데 문자를 입력했을 때 | 숫자 변환 실패 |
| CSV 파일의 데이터 형식이 잘못되었을 때 | 데이터 처리 실패 |
| 저장할 폴더가 없거나 권한이 없을 때 | 파일 저장 실패 |

이런 오류가 발생하면 프로그램이 바로 종료될 수 있습니다.  
예외 처리를 사용하면 오류가 발생해도 프로그램이 멈추지 않고 적절한 안내 메시지를 보여줄 수 있습니다.

자바의 `try-catch`와 비슷하게, 파이썬에서는 `try-except`를 사용합니다.

# 30. `try-except` 기본 구조

예외 처리는 다음과 같은 구조로 작성합니다.

```python
try:
    오류가 발생할 수 있는 코드
except 오류종류:
    오류가 발생했을 때 실행할 코드
```

`try` 블록에는 오류가 발생할 수 있는 코드를 작성합니다.  
`except` 블록에는 오류가 발생했을 때 처리할 코드를 작성합니다.


````python
try:
    number = int(input("숫자를 입력하세요: "))
    print("입력한 숫자:", number)
except ValueError:
    print("숫자로 변환할 수 없는 값입니다.")
````

# 31. `ValueError` 처리하기

`ValueError`는 값의 형태가 맞지 않을 때 자주 발생합니다.

예를 들어 `int("abc")`는 정수로 바꿀 수 없기 때문에 오류가 발생합니다.  
사용자 입력을 숫자로 변환할 때는 `try-except`를 사용하면 안전합니다.


````python
try:
    study_minutes = int(input("공부 시간을 분 단위로 입력하세요: "))
    print("공부 시간:", study_minutes, "분")
except ValueError:
    print("공부 시간은 숫자로 입력해야 합니다.")
````

# 34. CSV 파일 읽기와 예외 처리

CSV 파일을 읽을 때도 예외 처리를 함께 사용하면 좋습니다.

특히 다음 상황을 대비할 수 있습니다.

| 상황 | 처리 방법 |
|---|---|
| CSV 파일이 없음 | `FileNotFoundError` 처리 |
| 숫자 변환 실패 | `ValueError` 처리 |
| 예상하지 못한 오류 | `Exception` 처리 |

아래 예제는 학습 기록 CSV 파일을 안전하게 읽는 코드입니다.


````python
import csv

records = []

try:
    with open("study_records.csv", "r", encoding="utf-8") as file:
        reader = csv.reader(file)
        next(reader, None)

        for row in reader:
            date = row[0]
            subject = row[1]
            minutes = int(row[2])
            memo = row[3]

            records.append({
                "date": date,
                "subject": subject,
                "minutes": minutes,
                "memo": memo
            })

    print("CSV 파일을 불러왔습니다.")
    print(records)

except FileNotFoundError:
    print("저장된 CSV 파일이 없습니다. 새 데이터로 시작합니다.")
except ValueError:
    print("공부 시간 데이터가 숫자 형식이 아닙니다.")
except Exception as error:
    print("CSV 파일을 읽는 중 오류가 발생했습니다.")
    print("오류 내용:", error)
````

# 35. `else`와 `finally`

예외 처리에는 `else`와 `finally`를 함께 사용할 수도 있습니다.

| 구문 | 실행 시점 |
|---|---|
| `try` | 오류가 발생할 수 있는 코드 실행 |
| `except` | 오류가 발생했을 때 실행 |
| `else` | 오류가 발생하지 않았을 때 실행 |
| `finally` | 오류 발생 여부와 관계없이 항상 실행 |

처음에는 `try-except`만 사용해도 충분합니다.  
`else`와 `finally`는 프로그램 흐름을 더 명확하게 만들고 싶을 때 사용합니다.


````python
try:
    number = int(input("숫자 입력: "))
except ValueError:
    print("숫자를 입력해야 합니다.")
else:
    print("정상 입력:", number)
finally:
    print("입력 처리가 끝났습니다.")
````

# 36. 프로젝트에 예외 처리 적용하기

학습 기록 프로그램에서는 다음 부분에 예외 처리를 적용하면 좋습니다.

| 기능 | 예외 처리 대상 |
|---|---|
| 공부 시간 입력 | 숫자가 아닌 값을 입력한 경우 |
| 파일 불러오기 | 저장 파일이 없는 경우 |
| CSV 읽기 | 데이터 형식이 잘못된 경우 |
| 파일 저장 | 저장 중 오류가 발생한 경우 |

단, 모든 상황을 예외 처리로 해결하는 것은 아닙니다.

| 상황 | 처리 방식 |
|---|---|
| 메뉴 번호가 잘못됨 | `if`로 처리 |
| 검색 결과가 없음 | `if`로 처리 |
| 숫자 변환 실패 | `try-except`로 처리 |
| 파일이 없음 | `try-except`로 처리 |

즉, **예상 가능한 업무 규칙은 `if`**, **실행 중 오류 가능성은 `try-except`**로 처리하면 됩니다.


````python
def input_study_minutes():
    try:
        minutes = int(input("공부 시간(분): "))
        return minutes
    except ValueError:
        print("공부 시간은 숫자로 입력해야 합니다.")
        return None


minutes = input_study_minutes()

if minutes is not None:
    print("입력된 공부 시간:", minutes, "분")
else:
    print("공부 시간 입력에 실패했습니다.")
````

## 문제 4. CSV 파일 저장하기

다음 학생 데이터를 `students_practice.csv` 파일로 저장하세요.

```python
students = [
    {"name": "민수", "score": 90},

    {"name": "서준", "score": 95}
]
```

````python
# 여기에 코드를 작성하세요.
import csv

students = [
  {'name':'민수', 'score':90},
  {'name':'지훈', 'score':80},
  {'name':'서준', 'score':95}
]

with open('student_practice.csv', 'a', encoding='utf-8') as file:
  writer = csv.writer(file)
  writer.writerow(['name','score'])
  for i in students:
    content = []
    for j in i.values():
      content.append(j)
    writer.writerow(content)
````
