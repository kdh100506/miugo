# 파이썬 컬렉션 타입 정리

## 수업 목표

이번 시간에는 여러 개의 데이터를 하나로 묶어서 관리하는 **컬렉션 타입**을 복습하고 정리합니다.

파이썬에서 자주 사용하는 컬렉션 타입은 다음과 같습니다.

| 컬렉션 타입 | 특징 |
|---|---|
| 리스트 `list` | 여러 값을 순서대로 저장, 수정 가능 |

| 딕셔너리 `dict` | 키와 값을 한 쌍으로 저장 |
| 세트 `set` | 중복을 허용하지 않는 값들의 모음 |

# 1. 컬렉션 타입이란?

컬렉션 타입은 **여러 개의 값을 하나의 변수에 담아서 관리하는 자료형**입니다.

예를 들어 학생 5명의 점수를 저장한다고 생각해 봅니다.


````python
score1 = 80
score2 = 90
score3 = 70
score4 = 100
score5 = 85
````

위와 같이 변수 하나씩 따로 저장하면 관리하기 어렵습니다.  
이럴 때 리스트를 사용하면 여러 값을 하나로 묶을 수 있습니다.


````python
scores = [80, 90, 70, 100, 85]

print(scores)
````

# 2. 리스트 `list`

## 2-1. 리스트란?

리스트는 여러 개의 값을 **순서대로 저장하는 컬렉션 타입**입니다.  
대괄호 `[]`를 사용해서 만듭니다.


````python
students = ["민수", "지훈", "서준"]

print(students)
````

---

## 2-2. 리스트의 특징

| 특징 | 설명 |
|---|---|
| 순서가 있음 | 저장한 순서대로 관리됨(동일한 값 저장 가능) |
| 인덱스 사용 가능 | 위치 번호로 값을 꺼낼 수 있음 |
| 값 변경 가능 | 추가, 수정, 삭제 가능 |
| 여러 타입 저장 가능 | 숫자, 문자열, 불 자료형 등을 함께 저장 가능 |

---

## 2-3. 리스트 만들기


````python
numbers = [10, 20, 30, 40, 40]
names = ["민수", "지훈", "서준"]
mixed = ["민수", 17, True, 85.5]

print(numbers)
print(names)
print(mixed)
````

---

## 2-4. 리스트 인덱스

리스트는 각 값에 위치 번호가 있습니다.  
이 위치 번호를 **인덱스**라고 합니다.

파이썬의 인덱스는 `0`부터 시작합니다.


````python
fruits = ["사과", "바나나", "포도"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
````

---

## 2-5. 음수 인덱스

음수 인덱스를 사용하면 뒤에서부터 값을 가져올 수 있습니다.


````python
fruits = ["사과", "바나나", "포도"]

print(fruits[-1])
print(fruits[-2])
````

---

## 2-6. 리스트 값 수정하기

리스트는 저장된 값을 수정할 수 있습니다.


````python
scores = [80, 90, 70]

scores[2] = 100

print(scores)
````

---

## 2-7. 리스트에 값 추가하기

`append()`를 사용하면 리스트의 마지막에 값을 추가할 수 있습니다.


````python
students = ["민수", "지훈"]

students.append("서준")

print(students)
````

---

## 2-8. 원하는 위치에 값 추가하기

`insert()`를 사용하면 원하는 위치에 값을 추가할 수 있습니다.


````python
students = ["민수", "서준"]

students.insert(1, "지훈")

print(students)
````

---

## 2-9. 리스트 값 삭제하기

`remove()`는 값을 기준으로 삭제합니다.


````python
students = ["민수", "지훈", "서준"]

students.remove("지훈")

print(students)
````

`pop()`은 인덱스를 기준으로 삭제합니다.


````python
students = ["민수", "지훈", "서준"]

students.pop(1)

print(students)
````

---

## 2-10. 리스트 길이 구하기

`len()`을 사용하면 리스트에 몇 개의 값이 들어 있는지 알 수 있습니다.


````python
students = ["민수", "지훈", "서준"]

print(len(students))
````

---

## 2-11. 리스트와 반복문

리스트는 `for`문과 함께 자주 사용합니다.


````python
students = ["민수", "지훈", "서준"]

for student in students:
    print(student)
````

---

## 2-12. 리스트 합계와 평균 구하기


````python
scores = [80, 90, 70, 100]

total = 0

for score in scores:
    total += score

average = total / len(scores)

print("합계:", total)
print("평균:", average)
````

---

## 2-13. 리스트에서 자주 사용하는 함수

| 함수 | 설명 |
|---|---|
| `append()` | 마지막에 값 추가 |
| `insert()` | 원하는 위치에 값 추가 |
| `remove()` | 특정 값 삭제 |
| `pop()` | 특정 위치의 값 삭제 |
| `len()` | 길이 구하기 |
| `sum()` | 합계 구하기 |
| `max()` | 최댓값 구하기 |
| `min()` | 최솟값 구하기 |
| `sort()` | 정렬하기 |

예시 코드입니다.


````python
scores = [80, 90, 70, 100]

print(sum(scores))
print(max(scores))
print(min(scores))

scores.sort()
print(scores)
````

# 3. 튜플 `tuple`

## 3-1. 튜플이란?

튜플은 리스트처럼 여러 값을 순서대로 저장하지만, **한 번 만든 뒤 값을 수정할 수 없는 컬렉션 타입**입니다.  
소괄호 `()`를 사용해서 만듭니다.


````python
point = (10, 20)

print(point)
````

---

## 3-2. 튜플의 특징

| 특징 | 설명 |
|---|---|
| 순서가 있음 | 저장한 순서대로 관리됨 |
| 인덱스 사용 가능 | 위치 번호로 값을 꺼낼 수 있음 |
| 수정 불가능 | 값을 추가, 수정, 삭제할 수 없음 |
| 안전한 데이터 저장에 적합 | 변경되면 안 되는 값에 사용 |

---

## 3-3. 튜플 값 가져오기


````python
student = ("민수", 17, "1학년")

print(student[0])
print(student[1])
print(student[2])
````

---

## 3-4. 튜플은 수정할 수 없음

아래 코드는 오류가 발생합니다.


````python
student = ("민수", 17, "1학년")

student[1] = 18
````

튜플은 한 번 생성되면 값을 바꿀 수 없습니다.

---

## 3-5. 튜플을 사용하는 경우

튜플은 변경되면 안 되는 데이터를 저장할 때 사용합니다.

예를 들어 좌표값은 보통 쉽게 바뀌지 않는 하나의 묶음으로 볼 수 있습니다.


````python
position = (100, 200)

x = position[0]
y = position[1]

print("x좌표:", x)
print("y좌표:", y)
````

---

## 3-6. 튜플 언패킹

튜플에 들어 있는 값을 한 번에 여러 변수에 나눠 담을 수 있습니다.


````python
student = ("민수", 17)

name, age = student

print(name)
print(age)
````

# 4. 딕셔너리 `dict`

## 4-1. 딕셔너리란?

딕셔너리는 값을 순서가 아니라 **키와 값의 쌍**으로 저장하는 컬렉션 타입입니다.  
중괄호 `{}`를 사용해서 만듭니다.


````python
student = {
    "name": "민수",
    "age": 17,
    "score": 90
}

print(student)
````

---

## 4-2. 딕셔너리의 특징

| 특징 | 설명 |
|---|---|
| 키와 값으로 저장 | `"name": "민수"` 형태 |
| 키로 값 접근 | 인덱스 대신 키를 사용 |
| 값 수정 가능 | 기존 값을 바꿀 수 있음 |
| 데이터 표현에 적합 | 학생 정보, 상품 정보, 회원 정보 등 |

---

## 4-3. 딕셔너리 값 가져오기


````python
student = {
    "name": "민수",
    "age": 17,
    "score": 90
}

print(student["name"])
print(student["age"])
print(student["score"])
````

---

## 4-4. `get()`으로 값 가져오기

`get()`을 사용하면 키가 없을 때 오류 대신 기본값을 사용할 수 있습니다.


````python
student = {
    "name": "민수",
    "score": 90
}

print(student.get("name"))
print(student.get("age", "나이 정보 없음"))
````

---

## 4-5. 딕셔너리 값 수정하기


````python
student = {
    "name": "민수",
    "score": 90
}

student["score"] = 95

print(student)
````

---

## 4-6. 딕셔너리에 값 추가하기


````python
student = {
    "name": "민수",
    "score": 90
}

student["grade"] = "A"

print(student)
````

---

## 4-7. 딕셔너리 값 삭제하기

`del`을 사용해서 특정 키와 값을 삭제할 수 있습니다.


````python
student = {
    "name": "민수",
    "age": 17,
    "score": 90
}

del student["age"]

print(student)
````

---

## 4-8. 딕셔너리 반복문

딕셔너리의 키만 반복해서 가져올 수 있습니다.


````python
student = {
    "name": "민수",
    "age": 17,
    "score": 90
}

for key in student:
    print(key)
````

키와 값을 함께 가져오려면 `items()`를 사용합니다.


````python
student = {
    "name": "민수",
    "age": 17,
    "score": 90
}

for key, value in student.items():
    print(key, value)
````

---

## 4-9. 딕셔너리에서 자주 사용하는 함수

| 함수 | 설명 |
|---|---|
| `keys()` | 모든 키 가져오기 |
| `values()` | 모든 값 가져오기 |
| `items()` | 모든 키와 값 가져오기 |
| `get()` | 키로 값 가져오기 |
| `pop()` | 키를 기준으로 값 삭제 |

예시 코드입니다.


````python
student = {
    "name": "민수",
    "age": 17,
    "score": 90
}

print(student.keys())
print(student.values())
print(student.items())
````

# 5. 세트 `set`

## 5-1. 세트란?

세트는 **중복을 허용하지 않는 값들의 모음**입니다.  
중괄호 `{}`를 사용해서 만들지만, 딕셔너리와 다르게 키와 값의 쌍이 아닙니다.


````python
numbers = {1, 2, 3, 3, 4, 4, 5}

print(numbers)
````

중복된 값은 하나만 남습니다.

---

## 5-2. 세트의 특징

| 특징 | 설명 |
|---|---|
| 중복 허용 안 함 | 같은 값은 하나만 저장 |
| 순서가 없음 | 인덱스로 접근할 수 없음 |
| 값 추가와 삭제 가능 | `add()`, `remove()` 사용 |
| 중복 제거에 유용 | 리스트의 중복값 제거 가능 |

---

## 5-3. 세트에 값 추가하기


````python
numbers = {1, 2, 3}

numbers.add(4)

print(numbers)
````

---

## 5-4. 세트에서 값 삭제하기


````python
numbers = {1, 2, 3, 4}

numbers.remove(3)

print(numbers)
````

---

## 5-5. 세트로 중복 제거하기

리스트에 중복된 값이 있을 때 세트를 사용하면 중복을 제거할 수 있습니다.


````python
names = ["민수", "지훈", "민수", "서준", "지훈"]

unique_names = set(names)

print(unique_names)
````

다시 리스트로 바꿀 수도 있습니다.


````python
names = ["민수", "지훈", "민수", "서준", "지훈"]

unique_names = list(set(names))

print(unique_names)
````

## 5-6. 세트 연산

세트는 수학의 집합처럼 사용할 수 있습니다.


````python
python_students = {"민수", "지훈", "서준"}
java_students = {"지훈", "서준", "도윤"}

print(python_students & java_students)
print(python_students | java_students)
print(python_students - java_students)
print(python_students ^ java_students)
````

| 연산자 | 의미 |
|---|---|
| `&` | 교집합 |
| `\|` | 합집합 |
| `-` | 차집합 |
| `^`|대칭차집합|

# 9. 종합 예제: 학생 점수 관리 프로그램

아래 예제는 리스트와 딕셔너리를 함께 사용합니다.


````python
students = []

while True:
    print("===== 학생 점수 관리 =====")
    print("1. 학생 추가")
    print("2. 학생 목록 보기")
    print("3. 평균 점수 보기")
    print("0. 종료")

    menu = input("메뉴 선택: ")

    if menu == "1":
        name = input("이름: ")
        score = int(input("점수: "))

        student = {
            "name": name,
            "score": score
        }

        students.append(student)
        print("학생이 추가되었습니다.")

    elif menu == "2":
        for student in students:
            print(student["name"], student["score"])

    elif menu == "3":
        if len(students) == 0:
            print("등록된 학생이 없습니다.")
        else:
            total = 0

            for student in students:
                total += student["score"]

            average = total / len(students)
            print("평균 점수:", average)

    elif menu == "0":
        print("프로그램을 종료합니다.")
        break

    else:
        print("잘못된 메뉴입니다.")
````

# 11. 컬렉션 심화 학습

앞에서 배운 리스트, 튜플, 딕셔너리, 세트의 기본 사용법을 바탕으로  
실제 프로그램에서 자주 사용하는 기능과 실수하기 쉬운 부분을 조금 더 깊게 학습합니다.

## 심화 학습 목표

## 11-1. 리스트 슬라이싱

슬라이싱은 리스트의 일부 범위를 잘라서 **새 리스트**로 만드는 기능입니다.

```text
리스트[시작 인덱스 : 끝 인덱스 : 간격]
```

- 시작 인덱스는 포함합니다.
- 끝 인덱스는 포함하지 않습니다.
- 간격을 생략하면 기본값은 `1`입니다.
- 간격에 음수를 사용하면 뒤에서 앞으로 이동합니다.

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])    # 인덱스 1부터 3까지
print(numbers[:3])     # 처음부터 인덱스 2까지
print(numbers[3:])     # 인덱스 3부터 끝까지
print(numbers[::2])    # 두 칸 간격
print(numbers[::-1])   # 역순
````

### 슬라이싱에서 자주 틀리는 부분

```python
numbers[1:4]
```

위 표현은 인덱스 `1`, `2`, `3`의 값을 가져옵니다.  
끝 인덱스인 `4`는 포함하지 않습니다.

또한 슬라이싱은 일반적으로 새로운 리스트를 만들기 때문에 원본 리스트는 바뀌지 않습니다.

## 11-2. `sort()`와 `sorted()`의 차이

두 기능은 모두 데이터를 정렬하지만 동작 방식이 다릅니다.

| 구분 | `list.sort()` | `sorted()` |
|---|---|---|
| 종류 | 리스트 객체의 메서드 | 파이썬 내장 함수 |
| 원본 변경 | 변경함 | 변경하지 않음 |
| 반환값 | `None` | 정렬된 새 리스트 |
| 사용 대상 | 리스트 | 리스트, 튜플, 세트 등 반복 가능한 객체 |

````python
scores = [80, 100, 70, 90]

sorted_scores = sorted(scores)

print("원본:", scores)
print("새 리스트:", sorted_scores)

result = scores.sort()

print("정렬된 원본:", scores)
print("sort()의 반환값:", result)
````

> **주의:** `scores = scores.sort()`라고 작성하면 `scores`에는 정렬된 리스트가 아니라  
> `None`이 저장됩니다. `sort()`는 리스트 자체를 정렬하고 값을 반환하지 않기 때문입니다.

## 11-3. 리스트의 참조와 복사

파이썬 변수는 리스트 객체 자체가 아니라 객체를 가리키는 **참조**를 저장합니다.

따라서 다음과 같이 대입하면 두 변수가 같은 리스트를 가리킵니다.

````python
original = [10, 20, 30]
same_list = original

same_list.append(40)

print("original:", original)
print("same_list:", same_list)
print("같은 객체인가?", original is same_list)
````

서로 독립적인 리스트를 만들려면 `copy()` 또는 슬라이싱을 사용할 수 있습니다.

````python
original = [10, 20, 30]

copy1 = original.copy()
copy2 = original[:]

copy1.append(40)
copy2.append(50)

print("original:", original)
print("copy1:", copy1)
print("copy2:", copy2)
````

### 얕은 복사 주의

`copy()`는 바깥쪽 리스트만 새로 만듭니다.  
리스트 안에 또 다른 리스트가 들어 있는 경우 내부 리스트는 함께 공유될 수 있습니다.

````python
original = [[1, 2], [3, 4]]
copied = original.copy()

copied[0].append(100)

print("original:", original)
print("copied:", copied)
````

중첩 리스트까지 완전히 독립적으로 복사하려면 `copy` 모듈의 `deepcopy()`를 사용합니다.

````python
from copy import deepcopy

original = [[1, 2], [3, 4]]
copied = deepcopy(original)

copied[0].append(100)

print("original:", original)
print("copied:", copied)
````

## 11-4. 컴프리헨션

컴프리헨션은 반복문을 사용해 컬렉션을 만드는 코드를 간결하게 표현하는 문법입니다.

### 리스트 컴프리헨션

```text
[저장할 값 for 변수 in 반복 가능한 객체 if 조건식]
```

````python
numbers = [1, 2, 3, 4, 5, 6]

squares = [number ** 2 for number in numbers]
even_squares = [number ** 2 for number in numbers if number % 2 == 0]

print(squares)
print(even_squares)
````

### 딕셔너리 컴프리헨션과 세트 컴프리헨션

````python
numbers = [1, 2, 2, 3, 3, 3, 4]

square_dict = {number: number ** 2 for number in numbers}
square_set = {number ** 2 for number in numbers}

print(square_dict)
print(square_set)
````

컴프리헨션은 짧은 변환이나 필터링에 적합합니다.  
조건이 복잡하거나 여러 단계의 처리가 필요하면 일반 `for`문이 더 읽기 좋습니다.

## 11-5. `enumerate()`로 인덱스와 값 함께 사용하기

`enumerate()`는 반복할 때 **인덱스와 값**을 함께 제공합니다.

````python
students = ["민수", "지훈", "서준"]

for number, student in enumerate(students, start=1):
    print(number, student)
````

`start=1`을 지정하면 번호를 1부터 시작할 수 있습니다.  
리스트의 실제 인덱스가 바뀌는 것은 아니며, 반복문에서 반환하는 번호만 1부터 시작합니다.

## 11-6. `zip()`으로 여러 컬렉션 함께 반복하기

`zip()`은 여러 컬렉션에서 같은 위치의 값을 묶어서 반환합니다.

````python
names = ["민수", "지훈", "서준"]
scores = [90, 80, 95]

for name, score in zip(names, scores):
    print(name, score)

students = [
    {"name": name, "score": score}
    for name, score in zip(names, scores)
]

print(students)
````

> `zip()`은 가장 짧은 컬렉션의 길이에 맞춰 반복을 종료합니다.

````python
names = ["민수", "지훈", "서준"]
scores = [90, 80]

print(list(zip(names, scores)))
````

## 11-8. 딕셔너리 안전하게 다루기

존재하지 않는 키를 대괄호로 조회하면 `KeyError`가 발생합니다.

- 반드시 존재하는 키: `dictionary[key]`
- 존재하지 않을 수도 있는 키: `dictionary.get(key, 기본값)`
- 여러 값을 한 번에 추가하거나 수정: `update()`
- 키가 없을 때만 기본값 생성: `setdefault()`

````python
student = {
    "name": "민수",
    "score": 90
}

print(student.get("age", 0))

student.update({
    "score": 95,
    "grade": "A"
})

student.setdefault("subjects", [])
student["subjects"].append("파이썬")

print(student)
````

### 딕셔너리 키의 조건

딕셔너리의 키는 실행 중에 값이 바뀌지 않는 **해시 가능한 객체**여야 합니다.

사용 가능한 대표적인 키:

- 정수
- 실수
- 문자열
- 튜플(튜플 내부의 값도 변경 불가능해야 함)

리스트, 딕셔너리, 세트는 키로 사용할 수 없습니다.

````python
locations = {
    (35.87, 128.60): "대구",
    (37.57, 126.98): "서울"
}

print(locations[(35.87, 128.60)])
````

## 11-10. 중첩 컬렉션 검색하기

리스트 안에 딕셔너리가 들어 있는 구조는 실제 프로그램에서 자주 사용합니다.  
조건문과 반복문을 함께 사용하면 원하는 데이터를 검색할 수 있습니다.

````python
students = [
    {"name": "민수", "score": 90, "subjects": ["파이썬", "데이터베이스"]},
    {"name": "지훈", "score": 80, "subjects": ["자바", "파이썬"]},
    {"name": "서준", "score": 95, "subjects": ["인공지능", "파이썬"]}
]

high_score_students = []

for student in students:
    if student["score"] >= 90:
        high_score_students.append(student["name"])

print(high_score_students)
````

같은 작업을 리스트 컴프리헨션으로 표현할 수도 있습니다.

````python
students = [
    {"name": "민수", "score": 90},
    {"name": "지훈", "score": 80},
    {"name": "서준", "score": 95}
]

high_score_students = [
    student["name"]
    for student in students
    if student["score"] >= 90
]

print(high_score_students)
````

## 11-11. 딕셔너리가 들어 있는 리스트 정렬하기

`sorted()`의 `key` 매개변수에 정렬 기준을 지정할 수 있습니다.

````python
students = [
    {"name": "민수", "score": 90},
    {"name": "지훈", "score": 80},
    {"name": "서준", "score": 95}
]

sorted_students = sorted(
    students,
    key=lambda student: student["score"],
    reverse=True
)

for student in sorted_students:
    print(student["name"], student["score"])
````

`lambda`는 이름 없이 한 줄로 작성하는 간단한 함수입니다.

```python
lambda student: student["score"]
```

위 표현은 학생 딕셔너리를 전달받아 점수를 반환하며, 이 점수를 정렬 기준으로 사용합니다.

## 11-12. `any()`와 `all()`

- `any()`는 하나라도 참이면 `True`를 반환합니다.
- `all()`은 모든 값이 참이어야 `True`를 반환합니다.

````python
scores = [85, 90, 100, 75]

print("100점 학생이 있는가?", any(score == 100 for score in scores))
print("모두 60점 이상인가?", all(score >= 60 for score in scores))
````

names = ["민수", "지훈", "민수", "서준", "지훈", "도윤"]

> 단순히 `list(set(names))`를 사용하면 기존 순서가 보장되지 않습니다.

````python
names = ["민수", "지훈", "민수", "서준", "지훈", "도윤"]
first = []

for i in names:
  if i in names and not i in first:
    first.append(i)

print(first)

# 여기에 코드를 작성하세요.
````
