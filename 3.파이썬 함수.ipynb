# 1. 함수가 필요한 이유

함수는 **특정 기능을 수행하는 코드 묶음**입니다.

같은 코드를 여러 번 반복해서 쓰면 코드가 길어지고 수정하기 어려워집니다.  
함수를 사용하면 코드를 기능별로 나누고, 필요할 때마다 다시 사용할 수 있습니다.

## 함수의 장점

| 장점 | 설명 |
|---|---|
| 코드 재사용 | 같은 코드를 여러 번 다시 쓰지 않아도 됨 |
| 코드 정리 | 기능별로 코드를 나눌 수 있음 |
| 수정 편리 | 한 곳만 수정하면 여러 곳에 반영됨 |
| 협업에 유리 | 팀원이 기능별로 나누어 작업하기 쉬움 |

## 함수 없이 반복해서 작성한 코드


```python
print("안녕하세요.")
print("파이썬 수업에 오신 것을 환영합니다.")

print("안녕하세요.")
print("파이썬 수업에 오신 것을 환영합니다.")

print("안녕하세요.")
print("파이썬 수업에 오신 것을 환영합니다.")
```

## 함수로 정리한 코드


```python
def say_hello():
    print("안녕하세요.")
    print("파이썬 수업에 오신 것을 환영합니다.")

say_hello()
say_hello()
say_hello()
```

# 8. 반환값

반환값은 함수가 실행된 후 결과를 돌려주는 값입니다.  
`return`을 사용합니다.

## 기본 형식


```python
def 함수이름():
    return 결과값
```

```python
def add(a, b):
    return a + b

result = add(10, 20)

print(result)
```

`return a + b`는 계산 결과를 함수 밖으로 돌려줍니다.  
반환된 값은 변수에 저장하거나 바로 출력할 수 있습니다.


```python
def get_square(number):
    return number * number

print(get_square(5))
```

# 9. `print()`와 `return`의 차이

`print()`는 화면에 출력만 합니다.  
`return`은 함수의 결과를 함수 밖으로 돌려줍니다.

처음에는 둘이 비슷해 보이지만, 프로젝트에서는 `return`이 매우 중요합니다.

## `print()`만 사용한 경우


```python
def add_print(a, b):
    print(a + b)

result = add_print(10, 20)

print("result:", result)
```

위 함수는 `print()`만 하고 `return`이 없기 때문에 `result`에는 `None`이 들어갑니다.

## `return`을 사용한 경우


```python
def add_return(a, b):
    return a + b

result = add_return(10, 20)

print("result:", result)
```

`return`을 사용하면 함수의 결과를 다른 계산에 다시 사용할 수 있습니다.


```python
def add(a, b):
    return a + b

total = add(10, 20) + add(30, 40)

print(total)
```

# 10. `return`을 만나면 함수는 종료됨

함수 안에서 `return`을 만나면 함수 실행이 끝납니다.  
`return` 아래에 있는 코드는 실행되지 않습니다.


```python
def test():
    print("첫 번째 출력")
    return
    print("두 번째 출력")

test()
```

위 코드에서 `"두 번째 출력"`은 실행되지 않습니다.

# 11. 조건문과 함수

조건문을 함수 안에서 사용할 수 있습니다.


```python
def check_pass(score):
    if score >= 60:
        return "합격"
    else:
        return "불합격"

result = check_pass(85)

print(result)
```

등급을 반환하는 함수 예시입니다.


```python
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

print(get_grade(95))
print(get_grade(82))
print(get_grade(55))
```

# 12. 반복문과 함수

반복문도 함수 안에서 사용할 수 있습니다.


```python
def print_numbers():
    for i in range(1, 6):
        print(i)

print_numbers()

#매개변수가 있는 함수를 만들고 호출해 보세요.
```

리스트의 합계를 구하는 함수입니다.


```python
def get_total(numbers):
    total = 0

    for number in numbers:
        total += number

    return total

scores = [80, 90, 70, 100]

print(get_total(scores))
```

# 13. 리스트를 매개변수로 전달하기

함수에 리스트를 전달할 수 있습니다.


```python
def print_students(students):
    for student in students:
        print(student)

names = ["민수", "지훈", "서준"]

print_students(names)
```

리스트의 평균을 구하는 함수입니다.


```python
def get_average(scores):
    total = 0

    for score in scores:
        total += score

    return total / len(scores)

scores = [80, 90, 70, 100]

average = get_average(scores)

print("평균:", average)
```

# 18. 여러 개의 값 반환하기

파이썬 함수는 여러 개의 값을 반환할 수 있습니다.


```python
def calculate(a, b):
    plus = a + b
    minus = a - b
    multiply = a * b

    return plus, minus, multiply

result = calculate(10, 3)

print(result)
```

여러 개의 값을 반환하면 튜플 형태로 반환됩니다.

```python
def calculate(a, b):
    plus = a + b
    minus = a - b
    multiply = a * b

    return plus, minus, multiply

p, m, mul = calculate(10, 3)

print("더하기:", p)
print("빼기:", m)
print("곱하기:", mul)
```

# 22. 함수 설계 연습

함수를 만들 때는 다음 순서로 생각하면 좋습니다.

1. 이 함수가 하는 일은 무엇인가?
2. 함수 이름은 무엇으로 할 것인가?
3. 필요한 입력값은 무엇인가?
4. 반환해야 할 결과값은 무엇인가?
5. 함수 안에서 어떤 처리를 할 것인가?

예를 들어 점수를 받아 등급을 반환하는 함수는 다음과 같이 설계할 수 있습니다.

| 항목 | 내용 |
|---|---|
| 함수 이름 | `get_grade` |
| 입력값 | 점수 |
| 처리 | 점수에 따라 A, B, C, D, F 결정 |
| 반환값 | 등급 문자열 |


```python
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"
    elif score >= 60:
        return "D"
    else:
        return "F"

grade = get_grade(88)

print(grade)
```

# 23. 메뉴형 프로그램과 함수

프로젝트에서는 메뉴형 프로그램을 만들 때 함수를 많이 사용합니다.

함수를 사용하지 않으면 모든 코드가 한 곳에 몰려서 복잡해집니다.  
함수를 사용하면 기능별로 코드를 나눌 수 있습니다.


```python
def show_menu():
    print("===== 메뉴 =====")
    print("1. 인사하기")
    print("2. 목표 보기")
    print("0. 종료하기")

def say_hello():
    print("안녕하세요.")

def show_goal():
    print("오늘의 목표는 함수 문법을 이해하는 것입니다.")

while True:
    show_menu()

    menu = input("메뉴 선택: ")

    if menu == "1":
        say_hello()
    elif menu == "2":
        show_goal()
    elif menu == "0":
        print("프로그램을 종료합니다.")
        break
    else:
        print("잘못된 메뉴입니다.")
```

```python
'''
students = []

student = {
    "name": "민수",
    "score": 90
}
'''


def student(name, score):
    students=[]
    for i,j in zip(name, score):
        student = {}
        student['name'] = i
        student['score'] = j
        students.append(student)
    return students

name = ['가','나','다']
score = [50, 90, 70]

student(name,score)
```

# 26. 오늘의 핵심 정리

| 개념 | 설명 |
|---|---|
| 함수 | 특정 기능을 수행하는 코드 묶음 |
| `def` | 함수를 만들 때 사용하는 키워드 |
| 함수 호출 | 만든 함수를 실행하는 것 |
| 매개변수 | 함수 안으로 값을 전달받는 변수 |
| 인자 | 함수를 호출할 때 실제로 전달하는 값 |
| 반환값 | 함수가 실행 후 돌려주는 결과 |
| `return` | 함수의 결과를 반환하고 함수를 종료 |
| 지역변수 | 함수 안에서 만든 변수 |
| 전역변수 | 함수 밖에서 만든 변수 |
| 기본값 매개변수 | 인자가 없을 때 사용할 기본값을 가진 매개변수 |
| 키워드 인자 | 매개변수 이름을 지정해서 전달하는 인자 |
