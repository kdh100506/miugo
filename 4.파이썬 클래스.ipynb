# 1. 클래스가 필요한 이유

지금까지 학생 정보는 딕셔너리로 표현할 수 있었습니다.


````python
student = {
    "name": "민수",
    "score": 90
}

print(student["name"])
print(student["score"])
````

딕셔너리는 간단한 데이터를 표현할 때 좋습니다.  
하지만 학생과 관련된 기능까지 함께 관리하려면 클래스가 더 적합할 수 있습니다.

예를 들어 학생은 다음과 같은 데이터와 기능을 가질 수 있습니다.

| 구분 | 예시 |
|---|---|
| 데이터 | 이름, 점수 |
| 기능 | 정보 출력, 등급 계산, 합격 여부 판단 |

클래스는 이처럼 **데이터와 기능을 하나로 묶는 문법**입니다.

# 2. 클래스와 객체

클래스는 객체를 만들기 위한 **설계도**입니다.  
객체는 클래스로 만든 **실제 데이터**입니다.

| 비유 | 프로그래밍 |
|---|---|
| 붕어빵 틀 | 클래스 |
| 붕어빵 | 객체 |
| 붕어빵 속 재료 | 속성 |
| 붕어빵의 동작 | 메서드 |

Object-Oriented Programming(OOP, 객체 지향 프로그래밍)은 객체를 중심으로 프로그램을 구성하는 방식입니다.  
이번 수업에서는 OOP 전체를 깊게 배우지 않고, 클래스의 기본 사용법만 경험합니다.

# 3. 클래스 기본 문법

파이썬에서 클래스는 `class` 키워드를 사용해서 만듭니다.

```python
class 클래스이름:
    코드
```

클래스 이름은 보통 첫 글자를 대문자로 작성합니다.


````python
class Student:
    pass

student = Student()

print(student)
````

`pass`는 아직 작성할 코드가 없을 때 사용하는 문법입니다.

# 4. 객체 만들기

클래스를 만든 뒤에는 클래스 이름 뒤에 괄호를 붙여 객체를 만들 수 있습니다.


````python
class Student:
    pass

student1 = Student()
student2 = Student()

print(student1)
print(student2)
````

`student1`과 `student2`는 같은 `Student` 클래스로 만들었지만 서로 다른 객체입니다.

# 5. 속성

속성은 객체가 가지고 있는 데이터입니다.  
학생 객체라면 이름, 점수, 학년 등이 속성이 될 수 있습니다.


````python
class Student:
    pass

student = Student()

student.name = "민수"
student.score = 90

print(student.name)
print(student.score)
````

위 코드에서 `student.name`과 `student.score`가 속성입니다.

# 6. 여러 객체에 서로 다른 속성 저장하기

같은 클래스로 만든 객체라도 각자 다른 값을 가질 수 있습니다.


````python
class Student:
    pass

student1 = Student()
student1.name = "민수"
student1.score = 90

student2 = Student()
student2.name = "지훈"
student2.score = 80

print(student1.name, student1.score)
print(student2.name, student2.score)
````

# 7. 메서드

메서드는 클래스 안에 만든 함수입니다.  
객체가 할 수 있는 기능이라고 생각하면 됩니다.


````python
class Student:
    def show_info(self):
        print("학생 정보입니다.")

student = Student()
student.show_info()
````

클래스 안의 메소드는 첫 번째 매개변수로 보통 `self`를 사용합니다.  
self가 키워드는 아니지만 파이썬 개발자들의 암묵적 약속이다.  
@staticmethod의 경우에는 self 매개변수가 없다.  
(@staticmethod -> 모든 객체기 공유하는 메소드)

# 8. 속성과 메서드 함께 사용하기

메서드 안에서 객체의 속성을 사용할 수 있습니다.


````python
class Student:
    def show_info(self):
        print("이름:", self.name)
        print("점수:", self.score)

student = Student()
student.name = "민수"
student.score = 90

student.show_info()
````

# 9. `self`란?

`self`는 **현재 객체 자기 자신**을 가리키는 이름입니다.


````python
class Student:
    def show_name(self):
        print(self.name)

student1 = Student()
student1.name = "민수"

student2 = Student()
student2.name = "지훈"

student1.show_name()
student2.show_name()
````

# 10. `__init__()` 생성자

객체를 만든 뒤에 속성을 하나씩 넣는 것은 불편합니다.

그래서 `__init__()`을 사용합니다.  
`__init__()`은 객체가 만들어질 때 자동으로 실행되는 특별한 메서드입니다.

보통 **생성자(Constructor)**라고 부릅니다. -> 생성 X, 초기화 역할을 함  
클래스 만들 때 넣지 않으면 자동으로 추가해줌.(자바와 똑같음)


````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

student = Student("민수", 90)

print(student.name)
print(student.score)
````

`Student("민수", 90)`으로 객체를 만들면 `__init__()`이 자동으로 실행됩니다.

# 12. 생성자와 메서드 함께 사용하기

속성은 `__init__()`에서 만들고, 기능은 메서드로 만들 수 있습니다.


````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def show_info(self):
        print("이름:", self.name)
        print("점수:", self.score)

student = Student("민수", 90)

student.show_info()
````

# 13. 합격 여부를 판단하는 메서드


````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_pass(self):
        return self.score >= 60

student = Student("지훈", 55)

if student.is_pass():
    print("합격입니다.")
else:
    print("불합격입니다.")
````

`is_pass()` 메서드는 점수가 60점 이상이면 `True`, 그렇지 않으면 `False`를 반환합니다.

````python
student = {
    "name": "민수",
    "score": 90
}

print(student["name"])
print(student["score"])
````

# 22. 객체가 다른 객체에게 영향을 주는 예제

공격 기능을 메서드로 만들 수 있습니다.


````python
class GameCharacter:
    def __init__(self, name, hp, attack):
        self.name = name
        self.hp = hp
        self.attack = attack

    def attack_target(self, target):
        target.hp -= self.attack
        print(self.name, "이/가", target.name, "을/를 공격했습니다.")
        print(target.name, "의 남은 체력:", target.hp)

hero = GameCharacter("용사", 100, 20)
monster = GameCharacter("슬라임", 50, 10)

hero.attack_target(monster)
````

이 예제에서는 `hero` 객체가 `monster` 객체의 체력을 줄입니다.

# 25. 종합 예제: 클래스 기반 학생 점수 관리 프로그램

아래 예제는 클래스, 함수, 리스트, 조건문, 반복문을 함께 사용합니다.


````python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def get_grade(self):
        if self.score >= 90:
            return "A"
        elif self.score >= 80:
            return "B"
        elif self.score >= 70:
            return "C"
        elif self.score >= 60:
            return "D"
        else:
            return "F"

    def show_info(self):
        print(self.name, self.score, self.get_grade())


def show_menu():
    print("===== 학생 관리 =====")
    print("1. 학생 추가")
    print("2. 학생 목록 보기")
    print("3. 평균 점수 보기")
    print("0. 종료")


def add_student(students):
    name = input("이름: ")
    score = int(input("점수: "))

    student = Student(name, score)
    students.append(student)

    print("학생이 추가되었습니다.")


def show_students(students):
    if len(students) == 0:
        print("등록된 학생이 없습니다.")
        return

    for student in students:
        student.show_info()


def show_average(students):
    if len(students) == 0:
        print("등록된 학생이 없습니다.")
        return

    total = 0

    for student in students:
        total += student.score

    average = total / len(students)

    print("평균 점수:", average)


students = []

while True:
    show_menu()

    menu = input("메뉴 선택: ")

    if menu == "1":
        add_student(students)
    elif menu == "2":
        show_students(students)
    elif menu == "3":
        show_average(students)
    elif menu == "0":
        print("프로그램을 종료합니다.")
        break
    else:
        print("잘못된 메뉴입니다.")
````
